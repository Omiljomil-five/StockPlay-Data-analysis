```python
import pandas as pd
import os
from functools import reduce

# 경로 설정
data_path = os.path.join("..", "data", "price")
output_path = os.path.join("..", "output")

# output 폴더가 없으면 자동 생성
os.makedirs(output_path, exist_ok=True)

# 1. 파일 불러오기
open_df = pd.read_csv(os.path.join(data_path, "open.csv"))
high_df = pd.read_csv(os.path.join(data_path, "high.csv"))
low_df = pd.read_csv(os.path.join(data_path, "low.csv"))
close_df = pd.read_csv(os.path.join(data_path, "close.csv"))
vwap_df = pd.read_csv(os.path.join(data_path, "vwap.csv"))
volume_df = pd.read_csv(os.path.join(data_path, "volume.csv"))

# 2. 공통 컬럼: ['date', 'symbol']
dfs = [open_df, high_df, low_df, close_df, vwap_df, volume_df]
for df in dfs:
    df['date'] = pd.to_datetime(df['date'])

# 3. merge
price_df = reduce(lambda left, right: pd.merge(left, right, on=['date','symbol'], how='outer'),
                  dfs)

price_df.columns = ['date', 'symbol', 'open', 'high', 'low', 'close', 'vwap', 'volume']
```
