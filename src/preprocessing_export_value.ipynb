```python
import pandas as pd

# 1. CSV 불러오기
df = pd.read_csv("../data/export_value.csv")

# 2. 날짜형식 변환
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# 3. 중복 제거
df = df.drop_duplicates(subset=["date", "symbol"])

# 4. 결측치 처리
# (필요 시 export_value가 없는 행 제거)
df = df.dropna(subset=["export_value"])

# 5. 데이터 타입 변환
df["export_value"] = pd.to_numeric(df["export_value"], errors="coerce")

# 6. 이상치 탐지 (간단한 통계 확인)
print(df["export_value"].describe())

# 7. 결측치 비율 확인
missing_ratio = df.isna().mean() * 100
print("[결측치 비율]")
print(missing_ratio)

# 8. 날짜별 심볼 수, 총합 등 집계 예시
summary = df.groupby("date").agg(
    symbol_count=("symbol", "nunique"),
    total_export=("export_value", "sum"),
    mean_export=("export_value", "mean")
).reset_index()

# 9. 저장
summary.to_csv("../output/export_value_clean.csv", index=False)

print("✅ 전처리 완료! ../output/export_value_clean.csv 저장됨.")
```

```python
import pandas as pd
import numpy as np

# 1. 데이터 불러오기
df = pd.read_csv("../data/export_value.csv")

# 2. 기본 타입 확인 및 변환
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["export_value"] = pd.to_numeric(df["export_value"], errors="coerce")

# 3. 결측치 및 비정상 데이터 확인
missing = df.isna().sum()
print("🧩 [결측치 개수]\n", missing)

# 4. 잘못된 날짜(파싱 실패, 결측) 탐지
invalid_dates = df[df["date"].isna()]
if not invalid_dates.empty:
    print("\n⚠️ 잘못된 날짜 데이터:\n", invalid_dates)

# 5. 음수 / 0 수출값 탐지
invalid_exports = df[df["export_value"] <= 0]
if not invalid_exports.empty:
    print("\n⚠️ 수출액이 0 이하인 데이터:\n", invalid_exports)

# 7. 중복 데이터 확인
duplicates = df[df.duplicated(subset=["date", "symbol"], keep=False)]
if not duplicates.empty:
    print("\n⚠️ 중복된 (date, symbol) 조합:\n", duplicates)

# 8. 수치형 통계 요약
print("\n📊 [요약 통계]")
print(df["export_value"].describe())

# 9. 정리: 문제가 없는 행만 유지하고 싶다면
clean_df = df[df["export_value"] > 0].drop_duplicates(subset=["date","symbol"])
clean_df.to_csv("../output/export_value_clean.csv", index=False)
```
